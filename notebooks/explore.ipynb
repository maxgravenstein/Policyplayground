# Flat tax + UBI: explore

Change one number in the config, run, and look. A full run takes about 10 minutes (the baseline is cached after the first run). Set `MTR = True` to also compute marginal tax rates (slower).

```python
from flatubi.config import load_config
from flatubi import run, outputs, solver, report

cfg = load_config('../config.yaml')
# Change one number here, for example:
# cfg['ubi']['adult_annual'] = 12000
# cfg['tax']['flat_rate'] = 0.35
SCENARIO = 'steady_state'   # steady_state | phase1 | custom
MTR = False
```

```python
res = run.run(cfg, scenario=SCENARIO, mtr=MTR)
tables = outputs.all_tables(res)
solve = solver.solve_rate(res)
print(f'Revenue-neutral rate: {solve.rate_pct:.2f}%')
```
